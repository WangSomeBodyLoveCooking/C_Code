# 从文本中读取文件信息

```python
import os
import glob
def read_data(file_dir):
    date_list = []
    filename = os.path.basename(os.path.abspath(file_dir)).split('.')
    with open(file_dir, 'rt', encoding='utf-8') as f, open(f'data/{filename[0]}_split.{filename[1]}', 'wt', encoding='utf-8') as w:
        while 1:
            str = f.read(100)
            if not str :break
            w.write(str+'\n')

    with open('data/pi_digits_100w_split.txt') as r:
        date = r.readlines()
        str = ''
        for i in date:
            str += i.strip()
        if '1222' in str:
            print('find it')
        else:
            print('not find it')


read_data('data/pi_digits_100w.txt')
```
