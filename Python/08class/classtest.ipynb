# 创建dog类
# 创建dog类的实例

```python
class Dog:
    """小狗类"""
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def sit(self):
        """模拟小狗收到命令时蹲下"""
        print(f'{self.name} is now sitting.')

    def roll_over(self):
        """模拟小狗收到命令时dagun"""
        print(f'{self.name} rolled over!')

my_dog = Dog('Willie', 6)
print(f'My dog\'s name is {my_dog.name}')
print(f'My dog\'s age is {my_dog.age}')

my_dog.sit()
my_dog.roll_over()

```

# Car类
## 给属性添加默认值
## 修改属性的值

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        # 给属性增加默认值
        self.odometer_reading = 0
        self.gas = 60
    
    def get_descaiptive_name(self):
        return f'{self.year} {self.make} {self.model}'.title()
    
    def read_odometer(self):
        print(f'This car has {self.odometer_reading} miles on it.')

    def update_odometer(self, mileage):
        if mileage < self.odometer_reading:
            print('Can not roll back an odometer!')
            return
        self.odometer_reading = mileage

    def increment_odometer(self, mileage):
        if mileage <= 0:
            print('invalid mileage')
            return
        self.odometer_reading += mileage
    
    def fill_gas_tank(self, gas_num):
        self.gas += gas_num

my_car = Car('BYD', '汉', '2025')
print(my_car.get_descaiptive_name())
my_car.read_odometer()

# 直接修改属性的值
my_car.odometer_reading = 10
my_car.read_odometer()
# 通过方法修改属性值
my_car.update_odometer(5)
my_car.update_odometer(20)
my_car.read_odometer()
my_car.increment_odometer(-1)
my_car.increment_odometer(100)
my_car.read_odometer()
```

# 类的继承
## 定义子类特有的属性和方法
## 重写父类的同名方法
## 将实例作为属性

```python
class Battery:
    def __init__(self, battery_size=75):
        self.battery_size = battery_size
    
    def describe_battery(self):
        print(f'This batter has a {self.battery_size} -kwh battery.')
    
class EleCar(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery = Battery()
    
    def fill_gas_tank(self):
        print('This car doesnot need a gas tank.')

my_elecar = EleCar('Byd', '汉', 2025)
print(my_elecar.get_descaiptive_name())
my_elecar.battery.describe_battery()

my_elecar.fill_gas_tank()
```
